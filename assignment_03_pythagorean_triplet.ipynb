```python
a, b, c = map(int, (input("Enter Three Sides: ")).split())

side_list = sorted([a, b, c])

def check_right_triangle():
    if side_list[0]**2 + side_list[1]**2 == side_list[2]**2:
        print("The given sides are of a Right Angled Triangle")
    else:
        print("The given sides do not form a Right Angled Triangle")

check_right_triangle()
```
